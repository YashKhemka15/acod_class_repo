## Creating and reading data

A pandas **DataFrame** is a two-dimensional table made up of rows and columns.

We can create one manually, although in practice we will usually read data from a file.

```python
list1 = (['Alabama', 10], ['Michigan', 20], ['Texas', 50])
columns = ['State', 'Population in millions']
df = pd.DataFrame(list1, columns=columns)
df
```

```python
list1 = (['Alabama', 'Michigan', 'Texas'],[10, 20, 50])
columns = ['State', 'Population in millions']
df = pd.DataFrame(list1).T
df.columns = columns
df
```

<div class="alert alert-warning">
<b>Refresher.</b>

## Selecting Data

Selecting data is one of the most common pandas operations.

We will focus on three common situations:

1. Selecting columns.
2. Selecting rows by position or label.
3. Selecting rows using conditions.

We will practise using the wine reviews DataFrame.

### 1. Selecting columns

A single column can be selected using square brackets.

```python
reviews['country']
```

Multiple columns can be selected by passing a list of column names.

```python
reviews[['country', 'province','points','price']]
```

### Exercise

### 2. Selecting rows with `loc` and `iloc`

`loc[]` selects using **labels**. (index)

`iloc[]` selects using **positions**. (row and column positions)

```python
# Select rows with index labels 10, 750 and 1200
reviews.loc[[10, 750, 1200], ['country', 'province','points','price']]
```

```python
# Select the first 5 rows and first 4 columns by position
reviews.iloc[:5, :4]
```

<div class="alert alert-warning">
<b>Remember.</b>
<code>loc</code> works with labels and conditions.  
<code>iloc</code> works with numerical positions.
</div>

### 3. Selecting rows using conditions

Boolean conditions allow us to keep only observations that satisfy a rule.

```python
reviews.loc[reviews["country"]=='France']
```

We can combine conditions using:

* `&` for **and**
* `|` for **or**

Each condition should be placed inside parentheses.

```python
reviews.loc[(reviews["country"]=='France') & (reviews["points"]>=95)]
```

`.isin()` is useful when we want to match several possible values.

```python
reviews.loc[reviews["country"].isin(['France', 'Italy'])]
```

## Inspecting Data Quality

Before analysing a dataset, we should check whether the data look reliable.

Useful questions include:

* Are values missing?
* Are there duplicated rows?
* Do numerical values have plausible ranges?
* How many unique values does each variable contain?
* Are the data types appropriate?

These checks do not automatically tell us what to do, but they help us identify potential problems.

```python
# How many missing values are in each column?
reviews.isnull().sum()
reviews.isnull().sum().sort_values(ascending=False)
```

```python
# What proportion of each column is missing?

(reviews.isnull().mean()*100).sort_values(ascending=False)
```

```python
# How many completely duplicated rows are there?
reviews.duplicated().sum()
```

```python
# How many unique values are in each column?
reviews.nunique()
```

```python
# Quick numerical summary
reviews.describe()
```

<div class="alert alert-warning">
<b>Important.</b>
Finding missing values does not automatically mean that we should drop them or replace them.

Before changing the data, ask:

<ul>
<li>Why might the value be missing?</li>
<li>How much data would we lose by dropping rows?</li>
<li>Would replacing the value introduce unrealistic information?</li>
<li>Could the missingness itself be informative?</li>
</ul>
</div>
